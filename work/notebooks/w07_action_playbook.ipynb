# Ranked Actions and Reason Codes

The ranked queue prioritizes content based on the baseline score.

Reason Code:
- HIGH_TRAFFIC – Content has high impressions and organic sessions.
- LOW_TRAFFIC – Content has low visibility and may require optimization.

Action:
- Improve SEO
- Review Content
- Monitor Performance

```python

import pandas as pd
from datasets import load_dataset
from google.colab import userdata

token = userdata.get("HF_TOKEN")

dataset = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    split="train",
    token=token,
    streaming=True
)

df = pd.DataFrame(dataset.take(1000))

# Baseline score
df["baseline_score"] = (
    0.5 * df["gsc_impressions"] +
    0.3 * df["gsc_clicks"] +
    0.2 * df["sessions_organic"]
)

# Reason code
df["reason_code"] = df["baseline_score"].apply(
    lambda x: "HIGH_TRAFFIC" if x > 30 else "LOW_TRAFFIC"
)

# Action
df["action"] = df["reason_code"].map({
    "HIGH_TRAFFIC": "Improve SEO",
    "LOW_TRAFFIC": "Review Content"
})

queue = df.sort_values("baseline_score", ascending=False)

queue.head(10)
```
