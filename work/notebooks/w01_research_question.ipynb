## Quick Look at the Data

The dataset contains **30,000 rows** and **44 columns**.

The average CTR is **0.5107**, and the median word count is **2,877 words**.

Most pages have a **downward trend (16,262 pages)**, followed by **stable (5,962)** and **up (4,388)**. This suggests that many pages may require refreshing.

The correlation between search volume and impressions is **0.001**, indicating almost no linear relationship. This suggests that multiple features together are needed to make good predictions, making machine learning a suitable approach.

```python
import pandas as pd

df = pd.read_csv("Flyrank/data/raw/content_refresh_anonymized.csv")

# Basic statistics
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

print("\nAverage CTR:", round(df["ctr"].mean(), 4))
print("Median Word Count:", df["word_count"].median())

print("\nTrend Direction Distribution:")
print(df["trend_direction"].value_counts())

print("\nCorrelation between Search Volume and Impressions:")
print(round(df["search_volume"].corr(df["impressions_90d"]), 3))
```
